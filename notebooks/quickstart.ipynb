## 3. Train on `empty_goal_close`

The easiest scenario: the ball starts next to an unguarded goal, so a shaped-reward PPO agent learns to score quickly.

```python
!python -m opengoalrl.scripts.train --config opengoalrl/configs/empty_goal_close.yaml
```

## 4. Evaluate the trained agent

Run evaluation episodes and write per-episode metrics to a CSV.

```python
!python -m opengoalrl.scripts.evaluate \
    --config opengoalrl/configs/empty_goal_close.yaml \
    --model models/empty_goal_close/ppo_opengoalrl \
    --output models/empty_goal_close/eval_metrics.csv
```

## 5. Plot the learning curve inline

Read the training metrics CSV and plot mean episode reward over timesteps.

```python
import csv
import matplotlib.pyplot as plt

timesteps, rewards = [], []
with open('models/empty_goal_close/training_metrics.csv', newline='') as f:
    for row in csv.DictReader(f):
        timesteps.append(float(row['timestep']))
        rewards.append(float(row['mean_reward']))

plt.figure(figsize=(8, 4))
plt.plot(timesteps, rewards, linewidth=1.5)
plt.xlabel('Timesteps')
plt.ylabel('Mean Episode Reward')
plt.title('empty_goal_close - PPO learning curve')
plt.grid(True, alpha=0.3)
plt.tight_layout()
plt.show()
```
